```python
vector = []
vector.append(8)
vector.append(9)
vector.append(11)
print(vector)
print(vector[0])
print(vector[1])
print(vector[2])
#Imprime el ultimo elemento.
print(vector[-1])
print(vector[-2])
print(vector[-3])
```

```python
#Arreglo de diferentes tipos de datos
vector2 = [3, 4.56, "Luis"]
print(vector2)
```

```python
#Insertar un elemento en la posición 1
vector2.insert(1, "Managua")
print(vector2)
```

```python
#Recorrer el vector
for item in vector2:
    print(item)
```

```python
#Recorrer el vector
for index in range(4):
    print(vector2[index])
```
